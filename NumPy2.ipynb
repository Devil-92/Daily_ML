### Expressing Conditional Logic as Array Operation

```python
xarr = np.array([1.1 , 1.2 , 1.3 , 1.4 , 1.5])
yarr = np.array([2.1 , 2.2 , 2.3 , 2.4 , 2.5])
```

```python
cond = np.array([1 , 0 , 1 , 0 , 1])
```

```python
r = [(x if c else y) for x , y , c in zip(xarr , yarr , cond)]
print(r)
```

```python
xarr = np.array([1.1 , 1.2 , 1.3 , 1.4 , 1.5])
yarr = np.array([2.1 , 2.2 , 2.3 , 2.4 , 2.5])
cond = np.array([1 , 0 , 1 , 0 , 1])
```

```python
r = np.where(cond , xarr , yarr)             # np.where(Condition , X (scalar or vector) , Y (scalar or vector))
print(r)
```

```python
r = np.where(xarr == 1.1)
```

```python
print(r)
```

```python
r = np.where(cond , xarr , 67)
r
```

### Random Work Stimulation

```python
import random
pos = 0
walk = [pos]
n_steps = 100
for _ in range(n_steps):
  if random.randint(0,1):
    step = 1 ;
  else:
    step = -1
  pos += step
  walk.append(pos)
```

```python
plt.plot(walk)
```

```python
print("Stimulating the same on Numpy")
```

```python
nsteps = 1000
```

```python
rng = np.random.default_rng(seed = 12345)
```

```python
draws = rng.integers(0,2, size = nsteps)
```

```python
print("Stimulating Many random Walks at once")
```

```python
n_walks = 5000
n_steps = 1000
```

```python
draws = rng.integers(0, 2 , (n_walks , n_steps))
```

```python
draws[:20 , :20]
```

```python
steps = np.where(draws == 1 , 1 , -1)
```

```python
walks = np.cumsum(steps , axis = 1)
```

```python
walks
```

```python
walks.max()
```

```python
walks.min()
```

```python
hits30 = (np.abs(walks)>=30).any(axis = 1)
```

```python
hits30
```

```python
hits30.sum()
```

```python
crossing_times = (np.abs(walks[hits30])>= 30).argmax(axis = 1)
```

```python
crossing_times
```

```python
crossing_times.mean()
```
