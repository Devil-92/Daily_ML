#### Boolean Indexing

```python
names = np.array(["Bob" , "Joe" , "Will" , "Bob" , "Will" , "Joe" , "Joe"])
data = np.array([[1,2] , [-1,4] , [-3, 5] , [6 ,7] , [7 ,8] , [8 ,9] , [10 , 11]])
```

```python
names
```

```python
data
```

```python
names == "Bob"
```

```python
data[names == "Bob"]
```

```python
data[names == "Bob" , 1:]
```

```python
data[names == "Bob" , 1]
```

```python
print("Boolean array must be same length as the array axis it is indexing")
```

```python
names != "Bob"
```

```python
data[~(names == "Bob")]
```

```python
data[names != "Bob"]
```

```python
mask = (names == "Bob") | (names == "Will")
data[mask]
```

```python
mask
```

```python
data[mask]
```

```python
print("Selecting data from an array by Boolean indexing and assigning the result to a new variable always create a copy of the data")
```

```python
print("In Case of NumPy We can only use (| and &) not (or and and)")
```

```python
data[data<0] = 0
data
```

```python
data[names != "Joe"] = 7
data
```

#### Fancy Indexing

```python
arr = np.zeros((8,4))
```

```python
for i in range(8):
  arr[i] = i
```

```python
arr
```

```python
arr[[4 , 3 , 5 , 7]]
```

```python
print("To select a subset of rows in particular order")
```

```python
arr[[-3,-2,-1,0]]
```

```python
arr = np.arange(40).reshape((8,5))
arr
```

```python
arr[[1,2,3,4],[1,2,3,4]]
```

```python
arr[[1,5,7,2]][:,[0,3,1,2]]           # Quite Fancy
```

```python
arr[[1,2,3,4],[1,2,3,4]] = 0
arr
```

```python
print("Fancy Indexing & Boolean Indexing make copies of the matrix rather than returning a view")
```

#### Transposing Arrays and Swapping Axes

```python
arr = np.arange(15).reshape((3,5))
arr
```

```python
arr.T
```

```python
np.dot(arr,arr.T)
```

```python
arr@arr.T
```

```python
print("Simple .T is a special Case of Swapping Axes")
```

```python
arr
```

```python
arr.swapaxes(0,1)
```

```python
print("Both transposing and swapping axes returns a view of the array")
```
