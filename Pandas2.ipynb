### Function Application and Mapping

```python
frame = pd.DataFrame(np.random.standard_normal((4 ,3)) ,
                     columns = list('ABC') ,
                     index = list('abcd'))
frame
```

```python
np.abs(frame)
```

```python
def f1(x):
  return x.max() - x.min()
```

```python
f1(frame)
```

```python
frame.apply(f1)
```

```python
frame
```

```python
frame.apply(f1 , axis = 1)
```

```python
frame.apply(f1 , axis = 'columns')
```

```python
print("The function passed to apply need not return a scalar value , it can also return a Series with multiple values")
```

```python
def f2(x):
  return pd.Series([x.min() , x.max()] , index = ['min' , 'max'])
```

```python
frame.apply(f2)
```

```python
print("Element wise Python functions can be used with map")
```

```python
def my_format(x):
  return f"{x:2f}"
```

```python
frame.map(my_format)
```

```python
frame['C'].apply(my_format)
```

```python
print("DataFrame.map() applies an element wise function across the entire DataFrame while DataFrame.apply() applies a function along an axis")
```
