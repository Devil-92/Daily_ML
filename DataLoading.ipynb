#### XML

```python
print("XML Supports heirarcial , nested data with metadata similar to HTML but more general")
```

```python
from lxml import objectify
```

```python
path = "/content/Performance_MNR.xml"
```

```python
with open(path) as f :
  parsed = objectify.parse(f)
root = parsed.getroot()
```

```python
print("lxml.objectify parse the file and get the reference to the root node of the XML file with getroot")
```

```python
print("root.INDICATOR returns a generator yeilding each <INDICATOR> XML element")
```

```python
print("For each record we populate the dict of tag names to data values")
```

```python
data = []
skip_fields = ["PARENT_SEQ", "INDICATOR_SEQ","DESIRED_CHANGE", "DECIMAL_PLACES"]
```

```python
for elt in root.INDICATOR :
  el_data = {}
  for child in elt.getchildren():
    if child.tag in skip_fields :
      continue
    el_data[child.tag] = child.pyval
  data.append(el_data)
```

```python
print("Lastly convert this list of Dict to Dataframe")
```

```python
perf = pd.DataFrame(data)
```

```python
perf.head()
```

### HDF5 Format

HDF5 is a respected file format intended for storing large quantities of scientific array
data. It is available as a C library, and it has interfaces available in many other lan‐
guages, including Java, Julia, MATLAB, and Python. The “HDF” in HDF5 stands for
hierarchical data format. Each HDF5 file can store multiple datasets and supporting
metadata. Compared with simpler formats, HDF5 supports on-the-fly compression
with a variety of compression modes, enabling data with repeated patterns to be
stored more efficiently. HDF5 can be a good choice for working with datasets that
don’t fit into memory, as you can efficiently read and write small sections of much
larger arrays.

```python
frame = pd.DataFrame({
    "a" : np.random.standard_normal(100)
})
```

```python
frame
```
