```python
import re
def clean(strings):
  result = []
  for val in strings:
    val = val.strip()                       # Strip Starting Ending WhiteSpaces
    val = re.sub("[!#?]" , "" , val)        # Replace Special Characters
    val = val.title()                       # Make the String in All Small execpt string[0]
    result.append(val)
  return result
```

```python
clean(states)
```

```python
states = [" Alabama ", "Georgia!  ", "Georgia", "georgia", "FlOrIda","south carolina##", "West virginia?"]
```

```python
def clean_punc(string):
  return re.sub("[!?#]" , "" , string)

ops = [str.strip , clean_punc , str.title]
def clean_string(strings , ops):
  result = []
  for s in strings :
    for op in ops :
      s = op(s)
    result.append(s)
  return result
```

```python
clean_string(states , ops)
```
